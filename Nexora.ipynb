```python
# ==============================================================================
# NEXORA SENTINEL -- COMPLETE ML PIPELINE
# Run in Google Colab with your datasets in the same directory (or /content/)
# ==============================================================================


# ############################################################################
# SETUP & IMPORTS
# ############################################################################

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.utils import resample
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, roc_curve, ConfusionMatrixDisplay
)

sns.set_theme(style="darkgrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 5)
print("All imports loaded.")



# ############################################################################
# MODEL 1 - PURCHASE INTENT PREDICTOR
# ############################################################################

shoppers = pd.read_csv("online_shoppers_intention.csv")
print(f"Shape: {shoppers.shape}")
shoppers.head(3)


purchase_counts = shoppers["Revenue"].value_counts()
print(purchase_counts)
print(f"\nPurchase rate: {purchase_counts[True] / len(shoppers):.1%}")

fig, ax = plt.subplots(figsize=(6,4))
purchase_counts.plot(kind="bar", color=["#4a9eff","#ff6b6b"], ax=ax, edgecolor="white")
ax.set_xticklabels(["No Purchase", "Purchase"], rotation=0)
ax.set_title("Purchase Class Distribution", fontweight="bold")
ax.set_ylabel("Sessions")
plt.tight_layout()
plt.show()


df = shoppers.copy()
le = LabelEncoder()
df["VisitorType"] = le.fit_transform(df["VisitorType"])
df["Month"]       = le.fit_transform(df["Month"])
df["Weekend"]     = df["Weekend"].astype(int)
df["Revenue"]     = df["Revenue"].astype(int)

features = [
    "PageValues", "BounceRates", "ExitRates",
    "ProductRelated", "ProductRelated_Duration",
    "VisitorType", "TrafficType", "Month", "Weekend"
]

X = df[features]
y = df["Revenue"]

# Balance via oversampling minority class
X_y = pd.concat([X, y], axis=1)
majority = X_y[X_y.Revenue == 0]
minority = X_y[X_y.Revenue == 1]
minority_up = resample(minority, replace=True, n_samples=len(majority), random_state=42)
balanced = pd.concat([majority, minority_up])
X_bal = balanced[features]
y_bal = balanced["Revenue"]

X_train, X_test, y_train, y_test = train_test_split(
    X_bal, y_bal, test_size=0.2, random_state=42
)
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc  = scaler.transform(X_test)

print(f"Train: {len(X_train):,} | Test: {len(X_test):,}")
print(f"Features: {features}")


models_intent = {
    "Logistic Regression": LogisticRegression(max_iter=500, random_state=42),
    "Random Forest":       RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1),
    "XGBoost":             XGBClassifier(n_estimators=200, eval_metric="logloss", random_state=42, verbosity=0),
}

intent_results = {}

for name, model in models_intent.items():
    Xtr = X_train_sc if name == "Logistic Regression" else X_train
    Xte = X_test_sc  if name == "Logistic Regression" else X_test
    model.fit(Xtr, y_train)
    proba = model.predict_proba(Xte)[:,1]
    preds = model.predict(Xte)
    auc   = roc_auc_score(y_test, proba)
    intent_results[name] = {"model": model, "proba": proba, "preds": preds, "auc": auc}
    print(f"  {name:25s} | AUC: {auc:.4f}")
    print(classification_report(y_test, preds, target_names=["No Purchase","Purchase"]))
    print("-"*60)


plt.figure(figsize=(8,5))
colors = ["#4a9eff","#ff9f43","#ee5a24"]
for (name, res), color in zip(intent_results.items(), colors):
    fpr, tpr, _ = roc_curve(y_test, res["proba"])
    plt.plot(fpr, tpr, label=f'{name} (AUC={res["auc"]:.3f})', color=color, lw=2)
plt.plot([0,1],[0,1],"k--", lw=1)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves -- Purchase Intent Models", fontweight="bold")
plt.legend()
plt.tight_layout()
plt.show()


best_intent = intent_results["XGBoost"]["model"]
importance_df = pd.DataFrame({
    "Feature":    features,
    "Importance": best_intent.feature_importances_
}).sort_values("Importance", ascending=True)

fig, ax = plt.subplots(figsize=(8,5))
ax.barh(importance_df["Feature"], importance_df["Importance"], color="#4a9eff", edgecolor="white")
ax.set_title("XGBoost Feature Importance -- Purchase Intent", fontweight="bold")
ax.set_xlabel("Importance Score")
plt.tight_layout()
plt.show()


# Demo: Sentinel output for a high-intent session
sample_session = pd.DataFrame([{
    "PageValues": 85.0, "BounceRates": 0.01, "ExitRates": 0.03,
    "ProductRelated": 12, "ProductRelated_Duration": 850.0,
    "VisitorType": 1, "TrafficType": 2, "Month": 5, "Weekend": 0
}])

intent_score = intent_results["XGBoost"]["model"].predict_proba(sample_session)[0][1]

print("=" * 50)
print("  NEXORA SENTINEL -- INTENT ANALYSIS")
print("=" * 50)
print(f"  Purchase Intent Score : {intent_score:.0%}")
if intent_score >= 0.70:
    print("  [HIGH] Human mandate required before purchase.")
elif intent_score >= 0.40:
    print("  [MODERATE] Standard verification proceeding.")
else:
    print("  [LOW] Monitoring session.")
print("=" * 50)



# ############################################################################
# MODEL 2 - FRAUD RISK DETECTOR
# ############################################################################

creditcard = pd.read_csv("creditcard.csv")
print(f"Shape: {creditcard.shape}")
print(f"Fraud rate: {creditcard['Class'].mean():.4%}")
creditcard.head(3)


class_counts = creditcard["Class"].value_counts()
print(class_counts)

fig, ax = plt.subplots(figsize=(6,4))
class_counts.plot(kind="bar", color=["#4a9eff","#ff6b6b"], ax=ax, edgecolor="white")
ax.set_xticklabels(["Legitimate", "Fraud"], rotation=0)
ax.set_title("Transaction Class Distribution", fontweight="bold")
ax.set_ylabel("Transactions")
plt.tight_layout()
plt.show()


cc = creditcard.copy()
sc2 = StandardScaler()
cc["Amount"] = sc2.fit_transform(cc[["Amount"]])
cc["Time"]   = sc2.fit_transform(cc[["Time"]])

# Undersample majority for faster training; keep all fraud
fraud   = cc[cc.Class == 1]
legit   = cc[cc.Class == 0].sample(n=len(fraud)*10, random_state=42)
balanced_cc = pd.concat([fraud, legit]).sample(frac=1, random_state=42)

X_cc_b = balanced_cc.drop("Class", axis=1)
y_cc_b = balanced_cc["Class"]

X_cc_tr, X_cc_te, y_cc_tr, y_cc_te = train_test_split(
    X_cc_b, y_cc_b, test_size=0.2, random_state=42, stratify=y_cc_b
)
print(f"Train: {len(X_cc_tr):,} | Test: {len(X_cc_te):,}")
print(f"Fraud in test: {y_cc_te.sum():,} ({y_cc_te.mean():.1%})")


fraud_model = XGBClassifier(
    n_estimators=300, max_depth=6, learning_rate=0.05,
    scale_pos_weight=1, eval_metric="auc", random_state=42, verbosity=0
)
fraud_model.fit(X_cc_tr, y_cc_tr)

fraud_proba = fraud_model.predict_proba(X_cc_te)[:,1]
fraud_preds = fraud_model.predict(X_cc_te)
fraud_auc   = roc_auc_score(y_cc_te, fraud_proba)

print(f"Fraud Detection AUC: {fraud_auc:.4f}")
print()
print(classification_report(y_cc_te, fraud_preds, target_names=["Legitimate","Fraud"]))


cm = confusion_matrix(y_cc_te, fraud_preds)
disp = ConfusionMatrixDisplay(cm, display_labels=["Legitimate","Fraud"])
fig, ax = plt.subplots(figsize=(6,5))
disp.plot(ax=ax, colorbar=False, cmap="Blues")
ax.set_title("Fraud Detection -- Confusion Matrix", fontweight="bold")
plt.tight_layout()
plt.show()


fpr_f, tpr_f, _ = roc_curve(y_cc_te, fraud_proba)
plt.figure(figsize=(7,5))
plt.plot(fpr_f, tpr_f, color="#ee5a24", lw=2, label=f"XGBoost (AUC={fraud_auc:.3f})")
plt.plot([0,1],[0,1],"k--", lw=1)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve -- Fraud Risk Model", fontweight="bold")
plt.legend()
plt.tight_layout()
plt.show()


# Demo: Sentinel fraud output
sample_tx = X_cc_te.iloc[[0]]
fraud_score_demo = fraud_model.predict_proba(sample_tx)[0][1]

print("=" * 50)
print("  NEXORA SENTINEL -- FRAUD ANALYSIS")
print("=" * 50)
print(f"  Fraud Risk Score      : {fraud_score_demo:.0%}")
if fraud_score_demo >= 0.70:
    print("  [HIGH RISK] Additional verification required.")
elif fraud_score_demo >= 0.40:
    print("  [ELEVATED] Enhanced monitoring active.")
else:
    print("  [LOW RISK] Standard flow.")
print("=" * 50)



# ############################################################################
# MODEL 3 - ACCESSIBILITY MANDATE RECOMMENDER
# ############################################################################

accessibility = pd.read_csv("Accessibility dataset.csv")
sda           = pd.read_csv("SDA participants data.csv")
print("Accessibility dataset:", accessibility.shape)
print("Unique disability types:", accessibility["Type_Of_Disability"].unique()[:10])
print()
print("SDA dataset:", sda.shape)
print("SDA columns:", list(sda.columns))


disability_totals = (
    accessibility.groupby("Type_Of_Disability")["Data"]
    .sum().sort_values(ascending=False).head(10)
)

fig, ax = plt.subplots(figsize=(10,5))
disability_totals.plot(kind="barh", ax=ax, color="#4a9eff", edgecolor="white")
ax.set_title("Top 10 Disability Types by Prevalence (WHO Data)", fontweight="bold")
ax.set_xlabel("Cumulative Prevalence Score")
ax.invert_yaxis()
plt.tight_layout()
plt.show()


MANDATE_RULES = {
    "hearing":   "Sign Language / Visual OTP",
    "vision":    "Voice Biometric / Audio OTP",
    "motor":     "Eye Tracking / Switch Access",
    "cognitive": "Simplified PIN / Guardian Approval",
    "general":   "Passkey / Biometric Fingerprint",
}

MANDATE_STRENGTH = {
    "Passkey / Biometric Fingerprint":     1.00,
    "Voice Biometric / Audio OTP":         0.95,
    "Sign Language / Visual OTP":          0.92,
    "Eye Tracking / Switch Access":        0.90,
    "Simplified PIN / Guardian Approval":  0.85,
}

def recommend_mandate(profile):
    if profile.get("vision_impairment"):
        method, conf = MANDATE_RULES["vision"], 0.95
    elif profile.get("hearing_impairment"):
        method, conf = MANDATE_RULES["hearing"], 0.92
    elif profile.get("motor_impairment"):
        method, conf = MANDATE_RULES["motor"], 0.90
    elif profile.get("cognitive_impairment"):
        method, conf = MANDATE_RULES["cognitive"], 0.88
    else:
        method, conf = MANDATE_RULES["general"], 0.98
    return {"recommended_method": method, "accessibility_confidence": conf}

# Demo table
example_profiles = [
    ({"hearing_impairment": True,  "vision_impairment": False, "motor_impairment": False, "cognitive_impairment": False}, "Hearing impairment"),
    ({"hearing_impairment": False, "vision_impairment": True,  "motor_impairment": False, "cognitive_impairment": False}, "Vision impairment"),
    ({"hearing_impairment": False, "vision_impairment": False, "motor_impairment": True,  "cognitive_impairment": False}, "Motor impairment"),
    ({"hearing_impairment": False, "vision_impairment": False, "motor_impairment": False, "cognitive_impairment": False}, "General user"),
]

print(f"{'Profile':<22} {'Recommended Method':<36} {'Confidence':>11}")
print("-" * 72)
for profile, label in example_profiles:
    r = recommend_mandate(profile)
    print(f"{label:<22} {r['recommended_method']:<36} {r['accessibility_confidence']:>10.0%}")



# ############################################################################
# MODEL 4 - NEXORA TRUST ENGINE
# ############################################################################

def compute_trust_score(intent_score, fraud_score, accessibility_confidence, mandate_method):
    W_INTENT  = 0.30
    W_FRAUD   = 0.40
    W_ACCESS  = 0.15
    W_VERIFY  = 0.15

    fraud_safety = 1.0 - fraud_score
    verify_score = MANDATE_STRENGTH.get(mandate_method, 0.80)

    raw = (
        W_INTENT * intent_score +
        W_FRAUD  * fraud_safety +
        W_ACCESS * accessibility_confidence +
        W_VERIFY * verify_score
    )
    trust = round(raw * 100, 1)

    if trust >= 80:
        decision = "APPROVED -- Human mandate confirmed."
    elif trust >= 55:
        decision = "REVIEW   -- Enhanced verification required."
    else:
        decision = "BLOCKED  -- Transaction flagged for manual review."

    return {
        "trust_score": trust,
        "intent_contribution":          round(W_INTENT * intent_score * 100, 1),
        "fraud_contribution":           round(W_FRAUD  * fraud_safety * 100, 1),
        "accessibility_contribution":   round(W_ACCESS * accessibility_confidence * 100, 1),
        "verification_contribution":    round(W_VERIFY * verify_score * 100, 1),
        "decision": decision,
    }


scenarios = [
    {"label": "Normal Purchase (returning user, low fraud)",  "intent_score": 0.87, "fraud_score": 0.04, "profile": {"hearing_impairment":False,"vision_impairment":False,"motor_impairment":False,"cognitive_impairment":False}},
    {"label": "High Fraud Risk (suspicious transaction)",     "intent_score": 0.71, "fraud_score": 0.92, "profile": {"hearing_impairment":False,"vision_impairment":False,"motor_impairment":False,"cognitive_impairment":False}},
    {"label": "Accessibility User (vision impairment)",       "intent_score": 0.65, "fraud_score": 0.15, "profile": {"hearing_impairment":False,"vision_impairment":True,"motor_impairment":False,"cognitive_impairment":False}},
]

print("=" * 70)
print("  NEXORA TRUST ENGINE -- SCENARIO ANALYSIS")
print("=" * 70)

for s in scenarios:
    acc = recommend_mandate(s["profile"])
    res = compute_trust_score(s["intent_score"], s["fraud_score"],
                               acc["accessibility_confidence"], acc["recommended_method"])
    print(f"\n  Scenario : {s['label']}")
    print(f"  Intent   : {s['intent_score']:.0%}  |  Fraud Risk: {s['fraud_score']:.0%}")
    print(f"  Mandate  : {acc['recommended_method']}")
    print(f"  -------------------------------------------------------")
    print(f"  TRUST SCORE : {res['trust_score']}/100")
    print(f"  {res['decision']}")
print("\n" + "=" * 70)


fig, axes = plt.subplots(1, 3, figsize=(15, 5))
bar_colors = ["#4a9eff","#ff6b6b","#ffd32a","#0be881"]
bar_labels  = ["Intent","Fraud Safety","Accessibility","Verification"]

for i, s in enumerate(scenarios):
    acc = recommend_mandate(s["profile"])
    res = compute_trust_score(s["intent_score"], s["fraud_score"],
                               acc["accessibility_confidence"], acc["recommended_method"])
    vals = [
        res["intent_contribution"],
        res["fraud_contribution"],
        res["accessibility_contribution"],
        res["verification_contribution"],
    ]
    ax = axes[i]
    bars = ax.bar(bar_labels, vals, color=bar_colors, edgecolor="white")
    ax.set_ylim(0, 45)
    ax.set_title(f"{s['label']}\nTrust Score: {res['trust_score']}/100", fontsize=8, fontweight="bold")
    ax.set_ylabel("Score Contribution" if i == 0 else "")
    ax.tick_params(axis="x", labelrotation=15, labelsize=8)
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.3,
                f"{val:.1f}", ha="center", va="bottom", fontsize=8)

plt.suptitle("Nexora Trust Engine -- Score Breakdown by Scenario", fontweight="bold")
plt.tight_layout()
plt.show()



# ############################################################################
# END-TO-END PIPELINE DEMO
# ############################################################################

def nexora_sentinel_pipeline(browsing_features, transaction_df, user_profile):
    # Model 1 -- Intent
    session_df   = pd.DataFrame([browsing_features])
    intent_score = intent_results["XGBoost"]["model"].predict_proba(session_df)[0][1]

    # Model 2 -- Fraud
    fraud_score = fraud_model.predict_proba(transaction_df)[0][1]

    # Model 3 -- Accessibility mandate
    acc_result = recommend_mandate(user_profile)

    # Trust Engine
    trust_result = compute_trust_score(
        intent_score, fraud_score,
        acc_result["accessibility_confidence"],
        acc_result["recommended_method"]
    )

    return {
        "intent_score":     round(intent_score, 4),
        "fraud_risk_score": round(fraud_score, 4),
        "mandate_method":   acc_result["recommended_method"],
        **trust_result
    }


# Run a demo
demo_browsing = {
    "PageValues": 120.0, "BounceRates": 0.005, "ExitRates": 0.02,
    "ProductRelated": 18, "ProductRelated_Duration": 1200.0,
    "VisitorType": 1, "TrafficType": 3, "Month": 5, "Weekend": 1
}
demo_profile = {
    "hearing_impairment": False, "vision_impairment": False,
    "motor_impairment": False,   "cognitive_impairment": False
}

verdict = nexora_sentinel_pipeline(demo_browsing, X_cc_te.iloc[[5]], demo_profile)

print("=" * 60)
print("  NEXORA SENTINEL -- FULL PIPELINE VERDICT")
print("=" * 60)
print(f"  Purchase Intent Score : {verdict['intent_score']:.0%}")
print(f"  Fraud Risk Score      : {verdict['fraud_risk_score']:.0%}")
print(f"  Mandate Method        : {verdict['mandate_method']}")
print(f"  -----------------------------------------")
print(f"  TRUST SCORE           : {verdict['trust_score']}/100")
print(f"  {verdict['decision']}")
print("=" * 60)
```

```python
print("\n" + "="*60)
print("MODEL 1: PURCHASE INTENT PREDICTOR")
print("="*60)

feature_cols = ['PageValues', 'BounceRates', 'ExitRates', 'VisitorType', 'TrafficType', 'Month']
X = shoppers[feature_cols].copy()
y = shoppers['Revenue'].copy()

le_visitor = LabelEncoder()
X['VisitorType_enc'] = le_visitor.fit_transform(X['VisitorType'])
le_month = LabelEncoder()
X['Month_enc'] = le_month.fit_transform(X['Month'])
X_model = X.drop(['VisitorType', 'Month'], axis=1)

X_train, X_test, y_train, y_test = train_test_split(
    X_model, y, test_size=0.2, random_state=42, stratify=y)

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced'),
    'Random Forest': RandomForestClassifier(n_estimators=200, random_state=42, class_weight='balanced', n_jobs=-1),
    'XGBoost': xgb.XGBClassifier(n_estimators=200, max_depth=5, learning_rate=0.1, random_state=42,
                                  scale_pos_weight=5.0, eval_metric='logloss', use_label_encoder=False)
}

intent_results = {}
for name, model in models.items():
    print(f"\n▶ Training {name}...")
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    cv_scores = cross_val_score(model, X_train_s, y_train, cv=cv, scoring='roc_auc')
    model.fit(X_train_s, y_train)
    y_pred = model.predict(X_test_s)
    y_prob = model.predict_proba(X_test_s)[:, 1]
    intent_results[name] = {
        'model': model, 'cv_auc': cv_scores.mean(), 'test_auc': roc_auc_score(y_test, y_prob),
        'accuracy': accuracy_score(y_test, y_pred), 'precision': precision_score(y_test, y_pred),
        'recall': recall_score(y_test, y_pred), 'f1': f1_score(y_test, y_pred), 'y_prob': y_prob
    }
    print(f"  CV AUC: {cv_scores.mean():.4f} | Test AUC: {roc_auc_score(y_test, y_prob):.4f}")
    print(f"  Acc: {accuracy_score(y_test, y_pred):.4f} | Prec: {precision_score(y_test, y_pred):.4f} | Rec: {recall_score(y_test, y_pred):.4f}")

best_intent_name = max(intent_results, key=lambda x: intent_results[x]['test_auc'])
best_intent_model = intent_results[best_intent_name]['model']
print(f"\n✓ BEST INTENT MODEL: {best_intent_name} (AUC: {intent_results[best_intent_name]['test_auc']:.4f})")
```

```python
print("\n" + "="*60)
print("MODEL 3: NEXORA TRUST ENGINE")
print("="*60)

class NexoraTrustEngine:
    ACCESSIBILITY_PROFILES = {
        'hearing_impairment': {'method': 'Sign', 'confidence': 0.95},
        'vision_impairment': {'method': 'Voice', 'confidence': 0.92},
        'motor_impairment': {'method': 'Eye Tracking', 'confidence': 0.88},
        'general_user': {'method': 'Passkey', 'confidence': 0.98},
        'cognitive_impairment': {'method': 'Simplified Voice', 'confidence': 0.85}
    }
    VERIFICATION_STRENGTH = {
        'Passkey': 0.95, 'Biometric': 0.93, 'Sign': 0.90, 'Voice': 0.88,
        'Eye Tracking': 0.85, 'Simplified Voice': 0.82, 'SMS': 0.70, 'Email': 0.65
    }

    def __init__(self, intent_model, fraud_model, intent_scaler, fraud_scaler, le_visitor, le_month):
        self.intent_model = intent_model
        self.fraud_model = fraud_model
        self.intent_scaler = intent_scaler
        self.fraud_scaler = fraud_scaler
        self.le_visitor = le_visitor
        self.le_month = le_month

    def intent_score(self, user_features):
        return self.intent_model.predict_proba(self.intent_scaler.transform(user_features))[0, 1] * 100

    def fraud_safety(self, tx_features):
        fraud_prob = self.fraud_model.predict_proba(self.fraud_scaler.transform(tx_features))[0, 1]
        return (1 - fraud_prob) * 100

    def accessibility_rec(self, profile):
        return self.ACCESSIBILITY_PROFILES.get(profile, self.ACCESSIBILITY_PROFILES['general_user'])

    def verification_strength(self, method):
        return self.VERIFICATION_STRENGTH.get(method, 0.70) * 100

    def compute_trust(self, intent, fraud, access_conf, verif):
        return 0.25*intent + 0.35*fraud + 0.20*access_conf + 0.20*verif

    def evaluate(self, user_features, tx_features, profile):
        intent = self.intent_score(user_features)
        fraud = self.fraud_safety(tx_features)
        access = self.accessibility_rec(profile)
        access_conf = access['confidence'] * 100
        method = access['method']
        verif = self.verification_strength(method)
        trust = self.compute_trust(intent, fraud, access_conf, verif)

        if trust >= 85 and fraud >= 80:
            decision = "APPROVED - Low risk, high trust"
        elif trust >= 60:
            decision = "MANDATE REQUIRED - Human verification needed"
        elif trust >= 40:
            decision = "ESCALATED REVIEW - Additional checks required"
        else:
            decision = "BLOCKED - High risk transaction"

        return {
            'intent_score': round(intent, 1),
            'fraud_safety': round(fraud, 1),
            'accessibility': profile,
            'method': method,
            'access_conf': round(access_conf, 1),
            'verif_strength': round(verif, 1),
            'trust_score': round(trust, 1),
            'decision': decision
        }

trust_engine = NexoraTrustEngine(best_intent_model, best_fraud_model, scaler, scaler_fraud, le_visitor, le_month)
print("✓ Trust Engine initialized")
```
